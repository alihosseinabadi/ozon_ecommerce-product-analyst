```python

df['recent_ratio'] = df['Количество добавлений в избранное, 30 дней'] / df['Количество добавлений в избранное, все время']
df['recent_ratio'] = df['recent_ratio'].replace([np.inf, -np.inf], np.nan)

trending_products = df[df['Количество добавлений в избранное, 30 дней'] > 100].nlargest(10, 'recent_ratio')
print("\n  Trending Products (High Recent Ratio):")
for i, row in trending_products.iterrows():
    ratio = (row['Количество добавлений в избранное, 30 дней'] / row['Количество добавлений в избранное, все время']) * 100
    print(f"    {row['Название товара'][:40]}...")
    print(f"    {row['Количество добавлений в избранное, 30 дней']:,} recent vs {row['Количество добавлений в избранное, все время']:,} total ({ratio:.1f}%)")

declining_products = df[df['Количество добавлений в избранное, все время'] > 1000].nsmallest(10, 'recent_ratio')
print("\n  Declining Popularity:")
for i, row in declining_products.iterrows():
    ratio = (row['Количество добавлений в избранное, 30 дней'] / row['Количество добавлений в избранное, все время']) * 100
    print(f"    {row['Название товара'][:40]}...")
    print(f"    {row['Количество добавлений в избранное, 30 дней']:,} recent vs {row['Количество добавлений в избранное, все время']:,} total ({ratio:.1f}%)")
```

```python

category_stats = df.groupby('Категория 1 уровня').agg({
    'Количество добавлений в избранное, все время': ['sum', 'mean', 'median'],
    'Количество добавлений в избранное, 30 дней': ['sum', 'mean'],
    'Название товара': 'count'
}).round(2)

category_stats.columns = [
    'total_favorites_all',      # sum of all time favorites
    'avg_favorites_all',        # mean of all time favorites
    'median_favorites_all',     # median of all time favorites
    'total_favorites_30d',      # sum of 30-day favorites
    'avg_favorites_30d',        # mean of 30-day favorites
    'product_count'             # number of products
]

category_stats['favorites_per_product'] = category_stats['total_favorites_all'] / category_stats['product_count']
category_stats['recent_favorites_per_product'] = category_stats['total_favorites_30d'] / category_stats['product_count']
category_stats['recent_ratio'] = (category_stats['total_favorites_30d'] / category_stats['total_favorites_all']) * 100
category_stats['recent_ratio'] = category_stats['recent_ratio'].replace([np.inf, -np.inf], np.nan).fillna(0)

category_stats = category_stats.sort_values('total_favorites_all', ascending=False)

print("\nCategory Performance Summary:")
print("-" * 60)
print(f"{'Category':<30} {'Products':<10} {'Total Fav':<15} {'Avg Fav':<10}")
print("-" * 60)

for cat in category_stats.index[:10]:
    print(f"{cat[:28]:<30} {category_stats.loc[cat, 'product_count']:>8,}  "
          f"{category_stats.loc[cat, 'total_favorites_all']:>12,}  "
          f"{category_stats.loc[cat, 'avg_favorites_all']:>8.1f}")

print("\n2. Category Opportunity Score:")
print("-" * 40)

category_stats['opportunity_score'] = (category_stats['avg_favorites_all'] / category_stats['product_count']) * 100
category_stats['opportunity_score'] = category_stats['opportunity_score'].replace([np.inf, -np.inf], np.nan).fillna(0)

top_opportunities = category_stats.nlargest(10, 'opportunity_score')

print("High Opportunity Categories (High avg favorites vs product count):")
print("-" * 50)
for cat in top_opportunities.index:
    if category_stats.loc[cat, 'product_count'] > 10:  # Only categories with at least 10 products
        print(f"  {cat}:")
        print(f"    Products: {category_stats.loc[cat, 'product_count']:,.0f}")
        print(f"    Avg Favorites: {category_stats.loc[cat, 'avg_favorites_all']:.1f}")
        print(f"    Opportunity Score: {category_stats.loc[cat, 'opportunity_score']:.2f}")
        print()

print("\n3. Trending Categories (Best 30-Day Performance):")
print("-" * 40)

category_stats['recent_engagement'] = category_stats['total_favorites_30d'] / category_stats['product_count']
category_stats['recent_engagement'] = category_stats['recent_engagement'].replace([np.inf, -np.inf], np.nan).fillna(0)

trending_categories = category_stats.nlargest(10, 'total_favorites_30d')

for cat in trending_categories.index:
    if category_stats.loc[cat, 'product_count'] > 5:
        print(f"  {cat}:")
        print(f"    Recent Favorites (30d): {category_stats.loc[cat, 'total_favorites_30d']:,.0f}")
        print(f"    All Time Favorites: {category_stats.loc[cat, 'total_favorites_all']:,.0f}")
        recent_pct = (category_stats.loc[cat, 'total_favorites_30d'] / category_stats.loc[cat, 'total_favorites_all']) * 100
        print(f"    Recent % of Total: {recent_pct:.1f}%")
        print()

print("\n4. Category Performance Matrix:")
print("-" * 40)

performance_matrix = category_stats[['product_count', 'total_favorites_all', 'avg_favorites_all', 
                                      'total_favorites_30d', 'recent_ratio']].copy()


performance_matrix['performance_rating'] = pd.cut(
    performance_matrix['avg_favorites_all'],
    bins=[0, 10, 50, 100, 500, float('inf')],
    labels=['Poor', 'Fair', 'Good', 'Very Good', 'Excellent']
)

print("\nCategory Performance Ratings:")
print("-" * 50)
rating_counts = performance_matrix['performance_rating'].value_counts()
for rating, count in rating_counts.items():
    print(f"  {rating}: {count} categories")

print("\nTop Performing Categories (Highest Avg Favorites):")
print("-" * 50)
top_performers = category_stats.nlargest(10, 'avg_favorites_all')
for cat in top_performers.index:
    if category_stats.loc[cat, 'product_count'] > 5:
        print(f"  {cat}: {category_stats.loc[cat, 'avg_favorites_all']:.1f} avg favorites "
              f"({category_stats.loc[cat, 'product_count']} products)")
```

```python

category_stats = df.groupby('Категория 1 уровня').agg({
    'Количество добавлений в избранное, все время': ['sum', 'mean', 'median'],
    'Количество добавлений в избранное, 30 дней': ['sum', 'mean'],
    'Название товара': 'count'
}).round(2)


category_stats.columns = [
    'total_favorites_all',
    'avg_favorites_all',
    'median_favorites_all',
    'total_favorites_30d',
    'avg_favorites_30d',
    'product_count'
]


category_stats['favorites_per_product'] = category_stats['total_favorites_all'] / category_stats['product_count']
category_stats['recent_engagement'] = category_stats['total_favorites_30d'] / category_stats['product_count']
category_stats['recent_ratio'] = (category_stats['total_favorites_30d'] / category_stats['total_favorites_all']) * 100
category_stats['recent_ratio'] = category_stats['recent_ratio'].replace([np.inf, -np.inf], np.nan).fillna(0)


category_stats = category_stats.sort_values('total_favorites_all', ascending=False)


print("\nCategory Performance Summary:")
print("-" * 80)
print(f"{'Category':<30} {'Products':>10} {'Total Fav':>15} {'Avg Fav':>12} {'Recent %':>10}")
print("-" * 80)

for cat in category_stats.index[:15]:
    print(f"{cat[:28]:<30} {category_stats.loc[cat, 'product_count']:>10,} "
          f"{category_stats.loc[cat, 'total_favorites_all']:>15,} "
          f"{category_stats.loc[cat, 'avg_favorites_all']:>12.1f} "
          f"{category_stats.loc[cat, 'recent_ratio']:>9.1f}%")


print("\n2. Category Opportunity Score:")
print("-" * 40)


category_stats['opportunity_score'] = (category_stats['avg_favorites_all'] / category_stats['product_count']) * 100
category_stats['opportunity_score'] = category_stats['opportunity_score'].replace([np.inf, -np.inf], np.nan).fillna(0)

top_opportunities = category_stats.nlargest(10, 'opportunity_score')

print("High Opportunity Categories (High avg favorites vs product count):")
print("-" * 50)
for cat in top_opportunities.index:
    if category_stats.loc[cat, 'product_count'] > 10:
        print(f"\n  {cat}:")
        print(f"    Products: {category_stats.loc[cat, 'product_count']:,.0f}")
        print(f"    Avg Favorites: {category_stats.loc[cat, 'avg_favorites_all']:.1f}")
        print(f"    Opportunity Score: {category_stats.loc[cat, 'opportunity_score']:.2f}")

print("\n3. Trending Categories (Best 30-Day Performance):")
print("-" * 40)

trending_categories = category_stats.nlargest(10, 'total_favorites_30d')

for cat in trending_categories.index:
    if category_stats.loc[cat, 'product_count'] > 5:
        recent_pct = (category_stats.loc[cat, 'total_favorites_30d'] / 
                     category_stats.loc[cat, 'total_favorites_all']) * 100
        print(f"\n  {cat}:")
        print(f"    Recent Favorites (30d): {category_stats.loc[cat, 'total_favorites_30d']:,.0f}")
        print(f"    All Time Favorites: {category_stats.loc[cat, 'total_favorites_all']:,.0f}")
        print(f"    Recent % of Total: {recent_pct:.1f}%")


print("\n4. Category Performance Ratings:")
print("-" * 40)

def get_performance_rating(avg_favorites):
    if avg_favorites >= 500:
        return 'Excellent'
    elif avg_favorites >= 100:
        return 'Very Good'
    elif avg_favorites >= 50:
        return 'Good'
    elif avg_favorites >= 10:
        return 'Fair'
    else:
        return 'Poor'

category_stats['performance_rating'] = category_stats['avg_favorites_all'].apply(get_performance_rating)

print("\nPerformance Rating Distribution:")
print("-" * 40)
rating_counts = category_stats['performance_rating'].value_counts()
for rating, count in rating_counts.items():
    pct = (count / len(category_stats)) * 100
    print(f"  {rating}: {count} categories ({pct:.1f}%)")

print("\nTop Performing Categories (Excellent):")
print("-" * 40)
excellent = category_stats[category_stats['performance_rating'] == 'Excellent'].nlargest(10, 'avg_favorites_all')
for cat in excellent.index:
    print(f"  {cat}: {category_stats.loc[cat, 'avg_favorites_all']:.1f} avg favorites "
          f"({category_stats.loc[cat, 'product_count']} products)")
```

```python

category_stats = df.groupby('Категория 1 уровня').agg({
    'Количество добавлений в избранное, все время': ['sum', 'mean', 'median'],
    'Количество добавлений в избранное, 30 дней': ['sum', 'mean'],
    'Название товара': 'count'
}).round(2)

category_stats.columns = [
    'total_favorites_all',
    'avg_favorites_all',
    'median_favorites_all',
    'total_favorites_30d',
    'avg_favorites_30d',
    'product_count'
]

category_stats['favorites_per_product'] = category_stats['total_favorites_all'] / category_stats['product_count']
category_stats['recent_engagement'] = category_stats['total_favorites_30d'] / category_stats['product_count']
category_stats['recent_ratio'] = (category_stats['total_favorites_30d'] / category_stats['total_favorites_all']) * 100
category_stats['recent_ratio'] = category_stats['recent_ratio'].replace([np.inf, -np.inf], np.nan).fillna(0)

category_stats = category_stats.sort_values('total_favorites_all', ascending=False)

print("\nCategory Performance Summary:")
print("-" * 80)
print(f"{'Category':<30} {'Products':>10} {'Total Fav':>15} {'Avg Fav':>12} {'Recent %':>10}")
print("-" * 80)

for cat in category_stats.index[:15]:
    print(f"{cat[:28]:<30} {category_stats.loc[cat, 'product_count']:>10,} "
          f"{category_stats.loc[cat, 'total_favorites_all']:>15,} "
          f"{category_stats.loc[cat, 'avg_favorites_all']:>12.1f} "
          f"{category_stats.loc[cat, 'recent_ratio']:>9.1f}%")


print("\n2. Category Opportunity Score:")
print("-" * 40)

# Score = (Avg Favorites / Product Count) * 100
category_stats['opportunity_score'] = (category_stats['avg_favorites_all'] / category_stats['product_count']) * 100
category_stats['opportunity_score'] = category_stats['opportunity_score'].replace([np.inf, -np.inf], np.nan).fillna(0)

top_opportunities = category_stats.nlargest(10, 'opportunity_score')

print("High Opportunity Categories (High avg favorites vs product count):")
print("-" * 50)
for cat in top_opportunities.index:
    if category_stats.loc[cat, 'product_count'] > 10:
        print(f"\n  {cat}:")
        print(f"    Products: {category_stats.loc[cat, 'product_count']:,.0f}")
        print(f"    Avg Favorites: {category_stats.loc[cat, 'avg_favorites_all']:.1f}")
        print(f"    Opportunity Score: {category_stats.loc[cat, 'opportunity_score']:.2f}")


print("\n3. Trending Categories (Best 30-Day Performance):")
print("-" * 40)

trending_categories = category_stats.nlargest(10, 'total_favorites_30d')

for cat in trending_categories.index:
    if category_stats.loc[cat, 'product_count'] > 5:
        recent_pct = (category_stats.loc[cat, 'total_favorites_30d'] / 
                     category_stats.loc[cat, 'total_favorites_all']) * 100
        print(f"\n  {cat}:")
        print(f"    Recent Favorites (30d): {category_stats.loc[cat, 'total_favorites_30d']:,.0f}")
        print(f"    All Time Favorites: {category_stats.loc[cat, 'total_favorites_all']:,.0f}")
        print(f"    Recent % of Total: {recent_pct:.1f}%")


print("\n4. Category Performance Ratings:")
print("-" * 40)


def get_performance_rating(avg_favorites):
    if avg_favorites >= 500:
        return 'Excellent'
    elif avg_favorites >= 100:
        return 'Very Good'
    elif avg_favorites >= 50:
        return 'Good'
    elif avg_favorites >= 10:
        return 'Fair'
    else:
        return 'Poor'

category_stats['performance_rating'] = category_stats['avg_favorites_all'].apply(get_performance_rating)

print("\nPerformance Rating Distribution:")
print("-" * 40)
rating_counts = category_stats['performance_rating'].value_counts()
for rating, count in rating_counts.items():
    pct = (count / len(category_stats)) * 100
    print(f"  {rating}: {count} categories ({pct:.1f}%)")

print("\nTop Performing Categories (Excellent):")
print("-" * 40)
excellent = category_stats[category_stats['performance_rating'] == 'Excellent'].nlargest(10, 'avg_favorites_all')
for cat in excellent.index:
    print(f"  {cat}: {category_stats.loc[cat, 'avg_favorites_all']:.1f} avg favorites "
          f"({category_stats.loc[cat, 'product_count']} products)")
```
